```python
import os
import json
import random


DATA_DIR = "../data/defects4j"


def filtering():

    filter_list = []

    for bug_name in sorted(os.listdir(DATA_DIR)):
        current_path = os.path.join(DATA_DIR, bug_name, "current_report.json")
        relevant_path = os.path.join(
            DATA_DIR, bug_name, "relevant_reports_timewindow_90.json"
        )
        if not os.path.isfile(current_path) or not os.path.isfile(relevant_path):
            continue

        with open(relevant_path, "r", encoding="utf-8") as f:
            data = json.load(f)
            if len(data) >= 3:
                filter_list.append(bug_name)

    return filter_list

def select_random(bug_list):
    return sorted(random.sample(bug_list, 30))


bug_list = filtering()
print(f"Total count: {len(bug_list)}")

random_bugs = select_random(bug_list)
random_bugs = ['Cli_1', 'Cli_28', 'Cli_30', 'Codec_10', 'Compress_11', 'Compress_15', 'Compress_2', 'Compress_30', 'Compress_4', 'Compress_46', 'Compress_8', 'JxPath_5', 'Lang_11', 'Lang_16', 'Lang_24', 'Lang_34', 'Lang_37', 'Lang_42', 'Lang_49', 'Lang_58', 'Lang_8', 'Math_11', 'Math_16', 'Math_20', 'Math_43', 'Math_58', 'Math_76', 'Math_83', 'Math_84', 'Math_86']
print("Selected random 30 bugs")
temp = {}
for bug in random_bugs:
    temp[bug] = []
print(temp)

```
